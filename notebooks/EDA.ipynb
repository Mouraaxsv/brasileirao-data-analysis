### 3.3 Valores ausentes

```python
missing_percent = (
    df_clean.isnull().sum() / len(df_clean) * 100
).sort_values(ascending=False)

missing_percent[missing_percent > 0]
```

```python
df_clean[df_clean["publico"].isnull()]["ano_campeonato"].value_counts().sort_index()
```

- Os valores ausentes na coluna `publico` estão concentrados principalmente entre 2003 e 2006, indicando uma limitação histórica na disponibilidade dos dados.

```python
df_clean[
    df_clean["gols_mandante"].isnull() |
    df_clean["gols_visitante"].isnull()
]
```

```python
df_clean = df_clean.dropna(
    subset=["gols_mandante", "gols_visitante"]
)
```

```python
df_clean.shape
```

### Resumo do tratamento de valores ausentes

- A base não apresentou linhas duplicadas.
- As estatísticas detalhadas das partidas apresentam grande quantidade de valores ausentes entre 2003 e 2017.
- Os valores ausentes das estatísticas não foram substituídos por zero, pois a ausência de informação não representa necessariamente a ocorrência de zero eventos.
- Foi identificada uma partida sem gols registrados nas colunas `gols_mandante` e `gols_visitante`.
- Como os gols são informações essenciais para as análises futuras, essa partida foi removida da base analítica.

## 4. Engenharia de Atributos

### 4.1 Colunas de data

```python
#Novas colunas de data

df_clean["ano"] = df_clean["data"].dt.year
df_clean["mes"] = df_clean["data"].dt.month
df_clean["dia"] = df_clean["data"].dt.day
```

```python
df_clean[["data", "ano", "mes", "dia"]].head()
```

### 4.2 Validação de consistência

```python
# Gols mandante negativos
(df_clean["gols_mandante"] < 0).sum()
```

```python

# Gols visitante negativos

(df_clean["gols_visitante"] < 0).sum()
```

```python
#publico negativo

(df_clean["publico"] < 0).sum()
```

```python
#rodada invalida

df_clean["rodada"].describe()
```

### 4.3 Times únicos

```python
#Quantos clubes existem

times = pd.concat([
    df_clean["time_mandante"],
    df_clean["time_visitante"]
]).unique()

len(times)
```

```python
sorted(times)
```

### 4.4 Padronização de texto

```python
#Remover espaços em branco no início e no final das strings

colunas_texto = [
    "time_mandante",
    "time_visitante",
    "tecnico_mandante",
    "tecnico_visitante",
    "estadio",
    "arbitro"
]

for coluna in colunas_texto:
    df_clean[coluna] = df_clean[coluna].str.strip()
```

### 4.5 Total de gols e resultado da partida

```python
# Coluna com o total de gols 

df_clean["total_gols"] = (
    df_clean["gols_mandante"] +
    df_clean["gols_visitante"]
)
```

```python
df_clean[
    ["gols_mandante",
     "gols_visitante",
     "total_gols"]
].head()
```

```python
def resultado(row):
    if row["gols_mandante"] > row["gols_visitante"]:
        return "Mandante"
    elif row["gols_mandante"] < row["gols_visitante"]:
        return "Visitante"
    else:
        return "Empate"

df_clean["resultado"] = df_clean.apply(resultado, axis=1)
```

```python
df_clean["resultado"].value_counts()
```

## 5. Exportação da Base Tratada

```python
# Salvar a base de dados limpa em um novo arquivo CSV

df_clean.to_csv(
    "../data/processed/brasileirao_clean.csv",
    index=False
)
```

### Decisões tomadas

- Foram adicionadas as colunas `ano`, `mes`, `dia`, `total_gols` e `resultado`.
- A base tratada foi salva em `data/processed/brasileirao_clean.csv`.
